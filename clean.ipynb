```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
fund_master = pd.read_csv("data/raw_data/01_fund_master.csv")
portfolio = pd.read_csv("data/raw_data/09_portfolio_holdings.csv")
```

```python
fund_master.head()
fund_master.info()

portfolio.head()
portfolio.info()
```

```python
fund_master = fund_master.drop_duplicates()

# Remove rows with missing scheme names
fund_master = fund_master.dropna(subset=["scheme_name"])

# Fill other missing values if appropriate
fund_master = fund_master.fillna("Unknown")
```

```python
portfolio.columns
print(portfolio.columns.tolist())
```

```python
portfolio = portfolio.drop_duplicates()

portfolio = portfolio.dropna(
    subset=["amfi_code", "sector", "weight_pct"]
)

portfolio["weight_pct"] = pd.to_numeric(
    portfolio["weight_pct"],
    errors="coerce"
)

portfolio = portfolio.dropna(subset=["weight_pct"])

portfolio["portfolio_date"] = pd.to_datetime(
    portfolio["portfolio_date"],
    errors="coerce"
)

portfolio.head()
```

```python
portfolio.to_csv(
    "data/processed_data/portfolio_holdings_clean.csv",
    index=False
)
```

```python
fund_master.to_csv(
    "data/processed_data/fund_master_clean.csv",
    index=False
)
```
