```python
from sqlalchemy import create_engine
import pandas as pd

engine = create_engine("sqlite:///bluestock_mf.db")

nav = pd.read_csv("../data/processed_data/nav_history_clean.csv")
txn = pd.read_csv("../data/processed_data/investor_transactions_clean.csv")
perf = pd.read_csv("../data/processed_data/scheme_performance_clean.csv")

nav.to_sql(
    "fact_nav",
    engine,
    if_exists="replace",
    index=False
)

txn.to_sql(
    "fact_transactions",
    engine,
    if_exists="replace",
    index=False
)

perf.to_sql(
    "fact_performance",
    engine,
    if_exists="replace",
    index=False
)

print("Database created successfully!")
```
